# 🧠 Conceitos Básicos: Como o Python Funciona?

Para entender o Python de forma simples, precisamos analisar três características principais da linguagem: o **nível**, a **forma de tradução** e o fato de ser **orientada a objetos**.

---

## 1. Linguagem de Alto Nível vs. Baixo Nível

O "nível" de uma linguagem indica se ela está mais próxima da língua humana ou do circuito elétrico do computador.

* **Baixo Nível:** Fala a "língua da máquina". O código é escrito pensando diretamente no hardware e na memória do computador. É muito difícil de ler e escrever, mas executa super rápido.
  * *Exemplo (Assembly - para mostrar "Olá"):*
    ```assembly
    mov edx, 4
    mov ecx, msg
    mov ebx, 1
    mov eax, 4
    int 0x80
    ```

* **Alto Nível (Python):** Fala a "língua dos humanos" (utiliza palavras simples em inglês). O computador cuida de toda a parte pesada de memória e hardware para você.
  * *Exemplo (Python - para mostrar "Olá"):*
    ```python
    print("Olá, Mundo!")
    ```

---

## 2. Linguagem Compilada vs. Interpretada

O computador só ententende 0s e 1s (código de máquina). Portanto, todo programa precisa ser traduzido.

* **Linguagens Compiladas (ex: C, C++):**
  * Um programa chamado **Compilador** traduz o código inteiro de uma vez só antes de rodar.
  * Gera um arquivo pronto (como um `.exe`).
  * **Vantagem:** Executa super rápido.
  * **Desvantagem:** Se mudar uma linha, precisa traduzir tudo de novo. O arquivo gerado para Windows não roda no Linux.

* **Linguagens Interpretadas (ex: Python):**
  * Um programa chamado **Interpretador** lê e executa o código **linha por linha** em tempo real.
  * **Vantagem:** O mesmo arquivo `.py` roda em qualquer computador (Windows, Linux, Mac). Fácil e rápido para testar.
  * **Desvantagem:** É um pouco mais lenta do que as linguagens compiladas.

---

## 3. Python é uma Linguagem Orientada a Objetos

Dizer que o Python é **Orientado a Objetos** significa que ele organiza o código tratando as coisas como se fossem "objetos" da vida real.

Um objeto reúne duas coisas:
1. **Informações** (dados/características).
2. **Ações** (coisas que ele sabe fazer).

### 💡 Exemplo Prático em Python
No Python, **tudo é um objeto**. Quando você cria um texto, ele não é apenas uma palavra, mas sim um "objeto de texto" que já vem com várias ações prontas para você usar:

```python
# 'texto' é um objeto que guarda uma palavra
texto = "senai"

# O objeto 'texto' possui ações prontas, como transformar em maiúsculas:
print(texto.upper())  # Saída: SENAI

## 📌 **Resumo Rápido**
### CaracterísticaComo funciona no Python?
- Nível      |   Alto Nível (fácil de ler e parecido com o inglês).
- Tradução   |   Interpretada (executa o código linha por linha).
- Estrutura  |   Orientada a Objetos (trata dados e ações como objetos integrados).
