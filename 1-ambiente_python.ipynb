# 🐍 Guia Prático: Configuração do Ambiente de Desenvolvimento

Este guia apresenta o passo a passo completo para instalar a linguagem Python, o editor de código Visual Studio Code (VS Code) e a extensão oficial do Python para uso nas aulas.

---

## 1. Instalação da Linguagem Python ()

1. Acesse o site oficial do Python em [python.org/downloads](https://www.python.org/downloads/).
2. Clique no botão amarelo **Download Python 3.x.x** (versão mais recente estável).
3. Localize o arquivo baixado na sua pasta de *Downloads* e dê um duplo clique para iniciar o instalador.

## 1.2 Instalação da Linguagem Python (PowerShell)
```bash
winget install Python.Python.3.14
```

> ⚠️ **ATENÇÃO (Passo Crítico):**  
> Na PRIMEIRA tela do instalador, marque obrigatoriamente a caixa de seleção:  
> **`[x] Add python.exe to PATH`** (ou *Adicionar Python ao PATH*).  
>  
> *Se essa caixa não for marcada, o terminal do Windows e o VS Code não conseguirão reconhecer os comandos do Python.*

4. Após marcar a caixa do PATH, clique em **Install Now**.
5. Aguarde a conclusão e clique em **Close**.

### 🔍 Validação da Instalação do Python
1. Pressione as teclas **`Windows + R`**, digite `cmd` e pressione **Enter** para abrir o Prompt de Comando.
2. Digite o comando abaixo e pressione Enter:
   ```bash
   python --version
   ```
3. Se o terminal retornar a versão (ex: `Python 3.14.x`), a instalação foi concluída com sucesso.

---

## 2. Instalação do Visual Studio Code (VS Code)

1. Acesse o site oficial do VS Code em [code.visualstudio.com](https://code.visualstudio.com/).
2. Clique no botão **Download for Windows**.
3. Abra o arquivo baixado para iniciar o assistente de instalação.
4. Aceite o contrato de licença e clique em **Avançar**.
5. Na tela de tarefas adicionais, marque as seguintes opções recomendadas:
   - `[x]` *Criar um atalho na área de trabalho*
   - `[x]` *Adicionar "Abrir com Code" ao menu de contexto do Explorer*
6. Clique em **Instalar** e, ao finalizar, clique em **Concluir**.

## 2.1 Instalação do Visual Studio Code (PowerShell)
```bash
winget install Microsoft.VisualStudioCode
```

---

## 3. Instalação da Extensão do Python no VS Code

1. Abra o **Visual Studio Code**.
2. Acesse o painel de Extensões clicando no ícone no menu lateral esquerdo (quatro blocos) ou pelo atalho **`Ctrl + Shift + X`**.
3. Na barra de pesquisa, digite: **`Python`**.
4. Localize a extensão oficial desenvolvida pela **Microsoft** (primeiro resultado com selo de verificado).
5. Clique no botão azul **Install**.


---

## 4. Testando o Ambiente (Primeiro Programa)

1. No VS Code, acesse o menu superior: **File > Open Folder...** (*Arquivo > Abrir Pasta...*).
2. Selecione uma pasta no seu computador para os exercícios (ex: `Curso_Python`).
3. No painel esquerdo (*Explorer*), clique no ícone de **New File** (Novo Arquivo) e crie um arquivo chamado **`hello.py`**.
4. Escreva o seguinte código:
   ```python
   print("Hello World!")
   ```
5. Salve o arquivo com **`Ctrl + S`**.
6. Para executar o programa:
   - Clique no ícone de **Play (▶)** no canto superior direito; **OU**
   - Abra o terminal integrado (**`Ctrl + '`**) e digite:
     ```bash
     python hello.py
     ```
7. A mensagem `"Hello World!"` será exibida no terminal.
